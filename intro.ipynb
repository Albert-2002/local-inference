## Local Inference

running a pre-trained AI model on your own computer or server, rather than sending data to a remote inference API.

##### Required packages
- huggingface_hub\[cli\]
- llama-cpp-python

**llama.cpp** is a lightweight C/C++ inference engine for running large language models locally, especially models in the GGUF format.

```bash
wget "https://github.com/ggml-org/llama.cpp/releases/download/b10809/llama-b10809-bin-ubuntu-x64.tar.gz"
```

[Download llama.cpp b10809](https://github.com/ggml-org/llama.cpp/releases/download/b10809/llama-b10809-bin-ubuntu-x64.tar.gz)


The huggingface_hub package provides Python functionality for interacting with the Hugging Face Hub. The [cli] extra additionally gives you the command-line interface.

llama-cpp-python provides Python bindings for llama.cpp. llama.cpp is the underlying C/C++ inference engine that efficiently runs LLMs locally, particularly quantized GGUF models.
